**~ Import Libraries**

```python
from transformers import pipeline
import random
```

**~ Load Model**

```python
print("Loading AI model... (wait 1-2 mins first time)")

generator = pipeline("text-generation", model="gpt2")

print("Model loaded successfully!")
```

**~ Define Story Elements**

```python
GENRES = ["fantasy", "sci-fi", "mystery", "horror", "romance"]

CHARACTERS = [
    "a young wizard",
    "an alien explorer",
    "a detective",
    "a haunted artist",
    "a lonely robot"
]

SETTINGS = [
    "in a magical forest",
    "on Mars",
    "in a dark city",
    "inside a haunted house",
    "in a futuristic world"
]
```

**~ Prompt Generator Function**

```python
def generate_prompt():
    genre = random.choice(GENRES)
    character = random.choice(CHARACTERS)
    setting = random.choice(SETTINGS)

    return f"Write a {genre} story about {character} {setting} with a creative twist and meaningful ending."
```

**~ Story Generator Function**

```python
def generate_story(prompt):

    result = generator(
        prompt,
        max_length=300,
        num_return_sequences=1,
        temperature=0.8,
        truncation=True
    )

    story = result[0]['generated_text']

    # Paragraph formatting
    story = story.replace(". ", ".\n\n")

    return story
```

**~ Run the App**

```python
print("=== AI Story Generator (FREE VERSION) ===")

choice = input("Do you want custom input? (yes/no): ")

if choice.lower() == "yes":
    user_prompt = input("Enter your story idea: ")
else:
    user_prompt = generate_prompt()

print("\nGenerated Prompt:\n", user_prompt)
print("\nGenerating story...\n")

story = generate_story(user_prompt)

print("=== STORY OUTPUT ===\n")
print(story)
```
